#### Interview Question 1: Train/Validation/Test Split for Temporal Data

You're building a machine learning model to predict whether Chase app users will complete a money transfer in the next 7 days, based on their recent app interaction patterns.
You have a large-scale dataset of user interactions from January 2023 to December 2024 stored in a data lake. 

Each row contains:

* user_id: Unique user identifier
* interaction_date: Date of interaction (string format: 'YYYY-MM-DD')
* session_duration, num_screens_viewed, last_login_days_ago: Behavioral features
* completed_transfer_next_7days: Target variable (1/0)

##### Part 1

How would you split this data into training, validation, and test sets? 

What's the key principle you need to follow and why?

##### Part 2
Review this PySpark code snippet and identify TWO critical errors.

What are the two major problems with this approach? How would you fix them?

```python
from pyspark.sql import SparkSession
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.classification import LogisticRegression

spark = SparkSession.builder.appName("ChaseTransferPrediction").getOrCreate()

# Load Chase app interaction data
df = spark.read.parquet('s3://chase-data/user_interactions/')

# Randomly sample 70/15/15 split for train/val/test
train_data, val_data, test_data = df.randomSplit([0.7, 0.15, 0.15], seed=42)

# Prepare features
feature_cols = ['session_duration', 'num_screens_viewed', 'last_login_days_ago']
assembler = VectorAssembler(inputCols=feature_cols, outputCol='features')

train_data = assembler.transform(train_data)
val_data = assembler.transform(val_data)

# Train model
lr = LogisticRegression(featuresCol='features', labelCol='completed_transfer_next_7days')
model = lr.fit(train_data)

# Evaluate on validation set
val_predictions = model.transform(val_data)
```
